```python
df = pd.read_csv('dataset_penjualan_kantin - dataset_penjualan_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
#baris terjual hanya 65#
```

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
# Clean 'harga_satuan' column: remove 'Rp', '.', and ',' then convert to integer
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace('Rp', '', regex=False).str.replace('.', '', regex=False).str.replace(',', '', regex=False)
df['harga_satuan'] = pd.to_numeric(df['harga_satuan'], errors='coerce').fillna(0).astype(int)
print(df.dtypes)
#jawab: menggunakan fillna karena tidak ada barang yang terjual #
```

```python
# Convert 'jumlah_terjual' to numeric and handle missing values
df['jumlah_terjual'] = pd.to_numeric(df['jumlah_terjual'], errors='coerce').fillna(0).astype(int)

laris = df[df['jumlah_terjual'] > 20] # filtering
urut = df.sort_values(by='jumlah_terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga_satuan'] * df['jumlah_terjual'] # kolom turunan
ringkasan = df.groupby('nama_produk')['total_pendapatan'].sum() # agregasi (gunakan nama_produk sebagai pengganti 'menu')
print(ringkasan)
#bisa karena itu mengambil dari data penjualan kantin sekolah#
```

```python
import pandas as pd

# Memuat data dari file CSV
filepath = '/content/sample_data/california_housing_train.csv'
df = pd.read_csv(filepath)

# Menampilkan 5 baris pertama
display(df.head())

# Menampilkan informasi ringkas tentang kolom dan tipe data
df.info()
```

```python
# Menampilkan statistik deskriptif
display(df.describe())
```

```python
# Cek nilai kosong
print("Jumlah nilai kosong per kolom:")
print(df.isnull().sum())

# Cek data duplikat
print("\nJumlah baris duplikat:", df.duplicated().sum())
```

```python
# Memfilter data
df_filtered = df[(df['housing_median_age'] <= 10) & (df['median_house_value'] > 200000)]
print(f"Jumlah data setelah difilter: {len(df_filtered)} baris")
display(df_filtered.head())
```

```python
# Membuat kategori usia rumah
def categorize_age(age):
    if age <= 15:
        return 'Baru (<= 15 tahun)'
    elif age <= 30:
        return 'Sedang (16-30 tahun)'
    else:
        return 'Lama (> 30 tahun)'

df['age_group'] = df['housing_median_age'].apply(categorize_age)

# Mengelompokkan dan menghitung rata-rata harga serta populasi
summary_df = df.groupby('age_group')[['median_house_value', 'population']].mean().reset_index()
display(summary_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur tema grafik
sns.set_theme(style="whitegrid")

# Membuat histogram harga rumah
plt.figure(figsize=(10, 6))
sns.histplot(df['median_house_value'], kde=True, bins=50, color='skyblue')
plt.title('Distribusi Nilai Median Rumah (Median House Value)')
plt.xlabel('Nilai Rumah ($)')
plt.ylabel('Frekuensi')
plt.show()
```
